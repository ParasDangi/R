```python
# Q2. [20 Marks] Load the given employee dataset and perform ABC Analysis based on employee salary. Calculate the salary contribution percentage and cumulative percentage, classify employees into A, B, and C categories, and display the classified dataset.

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employees.csv")

abc = df[['First Name', 'Salary']].copy()
abc = abc.dropna()
abc = abc.groupby('First Name', as_index=False)['Salary'].sum()
abc = abc.sort_values(by='Salary', ascending=False)

total_salary = abc['Salary'].sum()
print("Total Salary =", total_salary)

abc['Contribution %'] = (abc['Salary'] / total_salary) * 100
abc['Cumulative %'] = abc['Contribution %'].cumsum()

def classify(x):
    if x <= 80:
        return "A"
    elif x <= 95:
        return "B"
    else:
        return "C"

abc["ABC Class"] = abc["Cumulative %"].apply(classify)
print(abc)
```

```python
# Q2. [20 Marks] Using the employee dataset, create a visualization dashboard to analyze inventory performance containing Inventory Summary, ABC Classification Table, Category-wise Bar Chart, Pie Chart.

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("employees.csv")

total_employees = len(df)
total_salary = df["Salary"].sum()
average_salary = df["Salary"].mean()
maximum_salary = df["Salary"].max()

print("========== INVENTORY SUMMARY ==========")
print("Total Employees :", total_employees)
print("Total Salary    :", total_salary)
print("Average Salary  :", round(average_salary, 2))
print("Maximum Salary  :", maximum_salary)

abc = df[["First Name", "Salary", "Team"]].copy()
abc = abc.sort_values(by="Salary", ascending=False)
abc["Cumulative Salary"] = abc["Salary"].cumsum()

total = abc["Salary"].sum()
abc["Cumulative %"] = (abc["Cumulative Salary"] / total) * 100

def classify(x):
    if x <= 70:
        return "A"
    elif x <= 90:
        return "B"
    else:
        return "C"

abc["ABC Class"] = abc["Cumulative %"].apply(classify)

print("\n========== ABC CLASSIFICATION ==========")
print(abc)

category_salary = df.groupby("Team")["Salary"].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 6))
category_salary.plot(kind="bar")
plt.title("Team-wise Salary")
plt.xlabel("Team")
plt.ylabel("Total Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 8))
category_salary.plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("Salary Distribution by Team")
plt.ylabel("")
plt.show()
```

```python
# Q1. [10 Marks] Load the (Sample - Superstore.csv) and perform a comprehensive data quality assessment by identifying missing values, duplicate records, incorrect data types, and invalid entries in the dataset. Prepare a summary report of all identified issues.

import pandas as pd

df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")

print("Dataset Shape:", df.shape)

missing = df.isnull().sum()
missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": missing.values,
    "Missing %": ((missing.values / len(df)) * 100).round(2)
})

duplicate_count = df.duplicated().sum()

df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")

invalid_quantity = ((df["Quantity"] <= 0) | (df["Quantity"].isna())).sum()
invalid_sales = ((df["Sales"] < 0) | (df["Sales"].isna())).sum()
invalid_discount = ((df["Discount"] < 0) | (df["Discount"] > 1) | (df["Discount"].isna())).sum()
invalid_dates = (df["Ship Date"] < df["Order Date"]).sum()

valid_ship_modes = ["First Class", "Second Class", "Standard Class", "Same Day"]
invalid_ship_mode = (~df["Ship Mode"].isin(valid_ship_modes)).sum()

valid_segments = ["Consumer", "Corporate", "Home Office"]
invalid_segment = (~df["Segment"].isin(valid_segments)).sum()

valid_categories = ["Furniture", "Office Supplies", "Technology"]
invalid_category = (~df["Category"].isin(valid_categories)).sum()

summary_report = pd.DataFrame({
    "Issue": [
        "Missing Values", "Duplicate Records", "Invalid Order Dates",
        "Invalid Ship Dates", "Invalid Quantity", "Invalid Sales",
        "Invalid Discount", "Invalid Shipping Dates", "Invalid Ship Mode",
        "Invalid Segment", "Invalid Category"
    ],
    "Count": [
        df.isnull().sum().sum(), duplicate_count, df["Order Date"].isna().sum(),
        df["Ship Date"].isna().sum(), invalid_quantity, invalid_sales,
        invalid_discount, invalid_dates, invalid_ship_mode,
        invalid_segment, invalid_category
    ]
})

print("\n==========================================")
print("       DATA QUALITY SUMMARY REPORT")
print("==========================================")
print(summary_report)
```

```python
# Q2. [20 Marks] Calculate the Year-over-Year (YoY) growth percentage and Compound Annual Growth Rate (CAGR). Display the results in a table and visualize the YoY growth and revenue trend using suitable graphs.

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Year': [2021, 2022, 2023, 2024, 2025],
    'Revenue': [100000, 120000, 150000, 180000, 220000]
}
df = pd.DataFrame(data)

df['YoY Growth (%)'] = df['Revenue'].pct_change() * 100

start = df['Revenue'].iloc[0]
end = df['Revenue'].iloc[-1]
n = len(df) - 1
cagr = ((end / start) ** (1 / n) - 1) * 100

print(df)
print("CAGR =", round(cagr, 2), "%")

plt.figure(figsize=(8, 5))
plt.bar(df['Year'], df['YoY Growth (%)'], color='skyblue')
plt.xlabel('Year')
plt.ylabel('Growth (%)')
plt.title('Year-over-Year Growth')
plt.xticks(df['Year'])
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(df['Year'], df['Revenue'], marker='o', color='green')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.title(f'Revenue Trend - CAGR = {cagr:.2f}%')
plt.xticks(df['Year'])
plt.grid(True)
plt.show()
```

```python
# Q2. [20 Marks] Load the (sales - sales.csv) dataset and calculate the 3-period Exponential Moving Average (EMA) of Sales. Display the original Sales values and calculated EMA values, and visualize both using a line chart.

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sales - sales.csv")

yearly_sales = df.groupby("Year")["Global_Sales"].sum().reset_index()

yearly_sales["EMA_3"] = yearly_sales["Global_Sales"].ewm(
    span=3,
    adjust=False
).mean()

print("Sales and 3-Period EMA:")
print(yearly_sales[["Year", "Global_Sales", "EMA_3"]])

plt.figure(figsize=(12, 6))
plt.plot(
    yearly_sales["Year"],
    yearly_sales["Global_Sales"],
    marker="o",
    label="Actual Sales"
)
plt.plot(
    yearly_sales["Year"],
    yearly_sales["EMA_3"],
    marker="o",
    label="3-Period EMA"
)
plt.xlabel("Year")
plt.ylabel("Global Sales (Millions)")
plt.title("Global Sales with 3-Period Exponential Moving Average")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Q2. [20 Marks] Create regional sales dataset for 2025 & 2026. Calculate average sales using groupby(), plot side-by-side bar chart, and convert format using melt().

import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Region': [
        'North', 'North', 'North',
        'South', 'South', 'South',
        'East', 'East', 'East',
        'West', 'West', 'West'
    ],
    'Sales_2025': [
        45000, 48000, 47000,
        52000, 55000, 53000,
        40000, 42000, 41000,
        60000, 62000, 61000
    ],
    'Sales_2026': [
        50000, 52000, 51000,
        58000, 60000, 59000,
        45000, 47000, 46000,
        68000, 70000, 69000
    ]
}
df = pd.DataFrame(data)

average_sales = df.groupby('Region')[['Sales_2025', 'Sales_2026']].mean()
print("\nAverage Sales by Region:")
print(average_sales)

average_sales.plot(
    kind='bar',
    figsize=(10, 6)
)
plt.title("Average Sales by Region: 2025 vs 2026")
plt.xlabel("Region")
plt.ylabel("Average Sales")
plt.xticks(rotation=0)
plt.legend(title="Year")
plt.tight_layout()
plt.show()

melted_df = df.melt(
    id_vars='Region',
    value_vars=['Sales_2025', 'Sales_2026'],
    var_name='Year',
    value_name='Sales'
)
print("\nTransformed Dataset using melt():")
print(melted_df)
```

```python
# Q2. [20 Marks] Load Sample - Superstore.csv and perform Pareto Analysis based on Category and Profit.

import pandas as pd

df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")

data = df[['Category', 'Profit']]

category_profit = data.groupby('Category')['Profit'].sum().reset_index()

category_profit = category_profit.sort_values(
    by='Profit', ascending=False
).reset_index(drop=True)

category_profit['Cumulative_Profit'] = category_profit['Profit'].cumsum()
total_profit = category_profit['Profit'].sum()
category_profit['Cumulative_Percentage'] = (
    category_profit['Cumulative_Profit'] / total_profit
) * 100

category_profit['Within_80_Percent'] = (
    category_profit['Cumulative_Percentage'] <= 80
)

print("Pareto Analysis:")
print(category_profit)
```

```python
# Q2. [20 Marks] Create inventory dataset and prepare summary report (Total products, Total inventory value, Count per ABC category, Percentage contribution).

import pandas as pd

data = {
    'Product': [
        'Laptop', 'Mobile', 'Tablet', 'Monitor', 'Keyboard',
        'Mouse', 'Printer', 'Headphones', 'Webcam', 'Speaker'
    ],
    'Quantity': [10, 20, 15, 25, 50, 60, 12, 40, 30, 35],
    'Unit_Price': [50000, 25000, 20000, 12000, 1500,
                   800, 15000, 3000, 2500, 4000],
    'ABC_Category': ['A', 'A', 'A', 'B', 'C',
                     'C', 'B', 'C', 'C', 'B']
}
df = pd.DataFrame(data)

df['Inventory_Value'] = df['Quantity'] * df['Unit_Price']

total_products = df['Product'].nunique()
total_inventory_value = df['Inventory_Value'].sum()

category_count = df['ABC_Category'].value_counts().reindex(
    ['A', 'B', 'C'], fill_value=0
)
category_percentage = (category_count / total_products) * 100

summary = pd.DataFrame({
    'Category': ['A', 'B', 'C'],
    'Number_of_Items': category_count.values,
    'Percentage_Contribution': category_percentage.round(2).values
})

print("\nABC ANALYSIS SUMMARY REPORT")
print("===========================")
print("Total Products:", total_products)
print("Total Inventory Value:", total_inventory_value)
print("\nCategory-wise Summary:")
print(summary)
```

```python
# Q2. [20 Marks] Load Advertising.xlsx, develop Simple Linear Regression model to predict Sales using TV Advertising. Display summary and plot line.

import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt

df = pd.read_excel("Advertising (1).xlsx")

X = df[['TV']]
y = df['Sales']
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print(model.summary())

pred = model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(df['TV'], y, label='Actual Sales')
plt.plot(df['TV'], pred, label='Regression Line', color='red')
plt.xlabel("TV Advertising")
plt.ylabel("Sales")
plt.title("Simple Linear Regression: TV vs Sales")
plt.legend()
plt.show()
```

```python
# Q3. Develop a reusable Python function that performs Simple Linear Regression for any selected predictor variable using the Advertising dataset.

import pandas as pd
import statsmodels.api as sm

df = pd.read_excel("Advertising (1).xlsx")

def simple_linear_regression(data, predictor, target):
    X = data[[predictor]]
    y = data[target]
    X = sm.add_constant(X)
    model = sm.OLS(y, X).fit()
    
    print("====================================")
    print("Simple Linear Regression")
    print("====================================")
    print("\nPredictor Variable:", predictor)
    print("Target Variable:", target)
    print("\nRegression Coefficients:")
    print(model.params)
    print("\nR-squared:", model.rsquared)
    print("\nP-values:")
    print(model.pvalues)
    print("\nRegression Summary:")
    print(model.summary())
    return model

simple_linear_regression(df, "TV", "Sales")
```

```python
# Q4. Develop Simple Linear Regression model using Cars dataset to predict COMB (kWh/100 km) from engine power (kW). Plot actual data with regression line.

import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt

df = pd.read_csv("cars.csv")

X = df[['(kW)']]
y = df['COMB (kWh/100 km)']
X = sm.add_constant(X)

model = sm.OLS(y, X).fit()

print("Regression Coefficients:")
print(model.params)

print("\nR-squared value:")
print(model.rsquared)

print("\nP-values:")
print(model.pvalues)

print("\nRegression Summary:")
print(model.summary())

pred = model.predict(X)

plt.figure(figsize=(8, 5))
plt.scatter(df['(kW)'], y, label='Actual Data')
plt.plot(df['(kW)'], pred, label='Regression Line', color='red')
plt.xlabel('Engine Power (kW)')
plt.ylabel('COMB (kWh/100 km)')
plt.title('Engine Power vs Combined Energy Consumption')
plt.legend()
plt.show()
```
